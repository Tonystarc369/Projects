```python
class smartphone:
    def __init__(self,company,model,ram,storage,price):
        self.company=company
        self.model=model
        self.ram=ram
        self.storage=storage
        self.price=price
    
    def display_info(self):
        return f"{self.company} with {self.model} having {self.ram} , {self.storage} with the {self.price}"

phone1=smartphone('Blackberry','e2',16,128,30000)
phone2=smartphone('Real me','GT 7T',16,128,30000)

print(phone1)
print(phone2)

class Gadget(smartphone):
    def __init__(self,company,model,ram,storage,price,gadget):
        super().__init__(company,model,ram,storage,price)
        self.gadget=gadget
    def show(self):
        return f"Avilable {self.gadget} with the specifications {self.company} with {self.model} having {self.ram} , {self.storage} with the {self.price}"
    

gadget1=Gadget('Blackberry','e2',16,128,30000,'Smartphone')
gadget2=Gadget('Real me','GT 7T',16,128,30000,'Smartphone')

print(gadget1.show())
print(gadget2.show())

```

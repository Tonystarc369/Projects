```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print(self.name, self.salary)
```

```python
emp1 = Employee("Vinay", 80000)
```

```python
for i in range(8,1):
    print(i)
```

```python
emp1.name = "Vinay"
emp1.salary = 80000
```

```python
emp1.display()
```
